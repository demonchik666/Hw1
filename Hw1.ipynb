```python
#5 Створіть два одновимірних масиви розміром 5, заповніть їх випадковими цілими числами в діапазоні від 1 до 10 та виконайте на них поелементні операції додавання, віднімання та множення.
e1 = np.random.randint(1, 11, size = 5)
e2 = np.random.randint(1, 11, size = 5)
e_sum = e1 + e2
print("5. Сума двох масивів:\nПерший масив: ", e1, "\nДругий масив: ", e2, "\nСума: ", e_sum)
e_sub = e1 - e2
print("Різниця двох масивів:\nПерший масив: ", e1, "\nДругий масив: ", e2, "\nРізниця: ", e_sub)
e_mult = e1 * e2
print("Добуток двох масивів:\nПерший масив: ", e1, "\nДругий масив: ", e2, "\nДобуток: ", e_mult)
```

```python
#6 Створіть два вектори розміром 7, заповніть довільними числами та знайдіть їх скалярний добуток.
f1 = np.random.randint(1, 11, size=(7))
f2 = np.random.randint(1, 11, size=(7))
f_scalar = np.dot(f1, f2)
print("6. Скалярний добуток двох масивів:\nПерший масив: ", f1, "\nДругий масив: ", f2, "\nСкалярний добуток: ", f_scalar)
```

```python
#7 Створіть дві матриці розміром 2x2 та 2x3, заповніть їх випадковими цілими числами в діапазоні від 1 до 10 та перемножте їх між собою.
g1 = np.random.randint(1, 11, size=(2, 2))
g2 = np.random.randint(1, 11, size=(2, 3))
g_scalar = np.dot(g1, g2)
print("7. Скалярний добуток двох матриць:\nПерша матриця:\n", g1, "\nДруга матриця:\n", g2, "\nПоелементний добуток:\n", g_scalar)
```

```python
#8 Створіть матрицю розміром 3x3, заповніть її випадковими цілими числами в діапазоні від 1 до 10 та знайдіть її обернену матрицю.
h = np.random.randint(1, 10, size = (3, 3))
h_inv = np.linalg.inv(h)
print("8. Обернена матриця для випадкової матриці 3x3:\nВипадкова матриця:\n", h, "\nОбернена матриця:\n", h_inv)
```
